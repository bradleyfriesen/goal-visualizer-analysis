```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

# Project paths
PROJECT_DIR = Path.cwd()
DATA_DIR = PROJECT_DIR.parents[1] / "data" / "processed" / "nhl-tracking-pipeline"
print(PROJECT_DIR)
print(PROJECT_DIR.parents[1])


SEASON = "20252026"

# Load games
games = pd.read_parquet(DATA_DIR / "games" / f"{SEASON}.parquet")

games.head()
```

```python
for path in sorted(DATA_DIR.rglob("*.parquet")):
    print(f"{path.relative_to(DATA_DIR)}  ({path.stat().st_size / 1e6:.1f} MB)")
```

```python
for path in sorted(DATA_DIR.rglob("*.parquet")):
    if "tracking" in path.parts:
        continue

    df = pd.read_parquet(path)

    print(f"\n{'='*70}")
    print(path.relative_to(DATA_DIR))
    print(f"Shape: {df.shape}")
    print(df.dtypes)
    display(df.head(3))

    del df
```

```python
import pyarrow.dataset as ds

tracking_path = DATA_DIR / "tracking" / f"{SEASON}.parquet"

dataset = ds.dataset(tracking_path, format="parquet")

game_id = 2025020001

table = dataset.to_table(
    filter=ds.field("game_id") == game_id
)

tracking = table.to_pandas()
```

```python
print(tracking.shape)
print(tracking.dtypes)
display(tracking.head(20))
```

```python
# Which events are represented?
print(tracking["event_id"].nunique())

# How many frames?
print(tracking["frame_idx"].nunique())

# Which entities?
print(tracking["entity_type"].value_counts())

# How many unique players?
print(tracking["player_id"].nunique())

# Time range
print(tracking["seconds_to_goal"].min(),
      tracking["seconds_to_goal"].max())
```
